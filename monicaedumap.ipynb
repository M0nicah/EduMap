```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#loading the dataset
data=pd.read_csv("kenya_primary_schools_updated.csv")
data.head()
```

```python
#null values
data.isna().count()
```

```python
#trying to spot outliers.
data=data.rename(columns={"Total Teachers" : "Total_Teachers"})

q1=data["Total_Teachers"].quantile(0.25)
q3=data["Total_Teachers"].quantile(0.75)
iqr=q3-q1
print(iqr)
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
```

```python
outliers = data[(data["Total_Teachers"] < lower_bound) | (data["Total_Teachers"] > upper_bound)]

print("Number of outliers:", outliers.shape[0])
outliers[["Name_of_Sc", "TotalEnrol", "Total_Teachers"]].head()
```

```python
#	Districts with highest teacher shortages 
low_staff_per_district = data.groupby("District") ["Total_Teachers"].sum().sort_values(ascending=False).reset_index()
low_staff_per_district.head(10)
```

```python
top10_districts_high_staff = data.groupby("District") ["Total_Teachers"].sum().sort_values(ascending=False).reset_index()
top10_districts_high_staff.head(10)
```

```python
#checking missing values
data['PupilTeach'].isna().sum(), (data["PupilTeach"] <=0).sum()
data2 = data[data["PupilTeach"] > 0].copy()
```

```python
#	Districts or locations with the highest average student-to-teacher ratios (PupilTeach).
district_avg_ratio = data2.groupby("District") ["PupilTeach"].mean().round(2).sort_values(ascending=True).reset_index()
district_avg_ratio.columns = ["District", "Average Student-to-Teacher Ratio"]
district_avg_ratio.head(10)
```

```python
district_avg_ratio.tail(10)
```

```python
location_avg_ratio = data2.groupby(["Location", "Province"]) ["PupilTeach"].mean().round(2).sort_values(ascending=False).reset_index()
location_avg_ratio.columns = ["Location", "Province", "Average Student-to-Teacher Ratio"]
location_avg_ratio.head()
```

```python
#Districts with highest teacher shortages 
#Filter out invalid rows
data2 = data2[(data2["TotalEnrol"] > 0) & (data2["Total_Teachers"] > 0)].copy()

#Calculate teachers needed based on 40:1 ratio
teachers_required = (data2["TotalEnrol"] /40).round().astype(int)
teachers_required.head()

# Calculate teacher shortages with no negative shortage.
data2["Teacher_Shortage"] = (teachers_required - data2["Total_Teachers"]).clip(lower=0)

#Teacher shortage by district
district_teacher_shortage = data2.groupby("District") ["Teacher_Shortage"].sum().round().sort_values(ascending=False).reset_index()
district_teacher_shortage.columns = ["District","Teachers_Needed"]
district_teacher_shortage.head()
```

```python
#schools with low or zero teaching staff.
data2=data2.rename(columns={"Name_of_Sc": "School_Name"})

data2["low_teachers"] = data2["Total_Teachers"] <=10
data2.head()

#school name with low staff
schools_with_low_staff = data2.groupby("School_Name")["Total_Teachers"].count().sort_values(ascending=False).reset_index()
schools_with_low_staff.columns = ['School_Name', 'Total_Teachers']
#schools_with_low_staff.head()
```
